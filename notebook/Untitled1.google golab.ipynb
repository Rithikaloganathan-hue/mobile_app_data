```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
df = pd.read_excel("cleaned_mobile_app_data_filled_4th_9th.xlsx")

sns.set_style("whitegrid")

# 1. Bar Chart - Category count
df["Category"].value_counts().plot(kind="bar")
plt.title("Apps by Category")
plt.xlabel("Category")
plt.ylabel("Number of Apps")
plt.tight_layout()
plt.show()

# 2. Histogram - App Ratings
df["Rating"].plot(kind="hist", bins=5, edgecolor="black")
plt.title("Distribution of App Ratings")
plt.xlabel("Rating")
plt.show()

# 3. Scatter Plot - Usage vs Rating
plt.scatter(df["Daily_Usage_Hours"], df["Rating"])
plt.title("Daily Usage vs Rating")
plt.xlabel("Daily Usage Hours")
plt.ylabel("Rating")
plt.show()

# 4. Box Plot - Daily Usage
sns.boxplot(y=df["Daily_Usage_Hours"])
plt.title("Daily Usage Distribution")
plt.show()

# 5. Heatmap - Correlation
sns.heatmap(
    df[["Rating", "Daily_Usage_Hours",
        "Installs_Count", "Price_USD"]].corr(),
    annot=True, cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
